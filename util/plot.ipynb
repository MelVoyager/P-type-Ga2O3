N_O的电荷转变能级

```python

# 数据准备
data = {
    r"N$_{O(I)}$": [4.03232, 3.542],
    r"N$_{O(II)}$": [2.88782, 2.599],
    r"N$_{O(III)}$": [3.83319, 3.363]
}

# 图形参数设置
x_positions = [1, 2, 3]  # 每列在x轴上的位置
colors = ['#2F5597', '#ED7D31']  # 不同来源的颜色
labels = ['我们的计算结果', r'文献$^1$中的计算结果']  # 图例标签
column_width = 0.8  # 每列占据的宽度

plt.figure(figsize=(8, 6))
ax = plt.gca()

# 绘制横线
for idx, (defect, values) in enumerate(data.items()):
    x_center = x_positions[idx]
    x_min = x_center - column_width/4
    x_max = x_center + column_width/4
    
    # 绘制两条能级线
    for i, value in enumerate(values):
        ax.hlines(y=value, 
                #  xmin=x_min + 0.05 + i*0.4,  # 横向偏移避免重叠
                #  xmax=x_max - 0.05 - (1-i)*0.4,
                xmin=x_min,
                xmax=x_max,
                 colors=colors[i], 
                 linewidth=3, 
                 label=labels[i] if idx == 0 else "")

# 坐标轴设置
ax.set_xticks(x_positions)
ax.set_xticklabels(data.keys(), fontsize=12)
plt.ylabel(r"$\varepsilon(0/-)$ eV", fontsize=12)
plt.ylim(2.5, 4.2)  # 设置Y轴范围

# 添加辅助网格
ax.grid(axis='y', linestyle='--', alpha=0.7)

# 图例设置
legend = ax.legend(loc='upper right', 
                  frameon=True,
                  edgecolor='black',
                  fontsize=10)
legend.get_frame().set_facecolor('white')

# 添加数据标签
for idx, (defect, values) in enumerate(data.items()):
    x_center = x_positions[idx]
    for i, value in enumerate(values):
        ax.text(x_center - 0.15 + 0.3*i, 
               value + 0.05, 
               f"{value:.2f} eV", 
               ha='center',
               color=colors[i],
               fontsize=10)

# 美化边框
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

绘制Fe,Cu体系的费米能级位置图示

```python
import matplotlib.pyplot as plt

# 带隙边界
vbm = 2.095
cbm = 6.7428

# Cu 系统费米能级
cu_fermi = {
    "Cu_{Ga(I)}": 2.5483269,
    "Cu_{Ga(I)}-N_{O(I)}": 3.3027539,
    "Cu_{Ga(I)}-N_{O(II)}": 3.10956582,
    "Cu_{Ga(I)}-N_{O(III)}": 3.2246764,
    "Cu_{Ga(II)}": 2.48227186,
    "Cu_{Ga(II)}-N_{O(I)}": 3.28740886,
    "Cu_{Ga(II)}-N_{O(II)}": 2.99816108,
    "Cu_{Ga(II)}-N_{O(III)}": 3.19953939,
}

# Fe 系统费米能级
fe_fermi = {
    "Fe_{Ga(I)}": 4.24051661,
    "Fe_{Ga(I)}-N_{O(I)}": 2.73651925,
    "Fe_{Ga(I)}-N_{O(II)}": 2.69944739,
    "Fe_{Ga(I)}-N_{O(III)}": 2.87581599,
    "Fe_{Ga(II)}": 5.1957575,
    "Fe_{Ga(II)}-N_{O(I)}": 3.35306625,
    "Fe_{Ga(II)}-N_{O(II)}": 3.15798264,
    "Fe_{Ga(II)}-N_{O(III)}": 3.3337761
}

def plot_fermi_levels(fermi_dict, title):
    plt.figure(figsize=(6, 6))
    y_positions = range(len(fermi_dict))
    labels = list(fermi_dict.keys())
    values = list(fermi_dict.values())

    # 绘制费米能级点和文本
    for y, val, label in zip(y_positions, values, labels):
        plt.plot(val, y, 'o', color='blue')
        plt.text(val + 0.05, y, f"{val:.2f} eV", va='center', fontsize=8)

    # 绘制 VBM / CBM 线
    plt.axvline(vbm, color='green', linestyle='--', label=f'VBM = {vbm:.2f} eV')
    plt.axvline(cbm, color='red', linestyle='--', label=f'CBM = {cbm:.2f} eV')

    # 填充价带左侧和导带右侧区域
    plt.axvspan(vbm - 0.5, vbm, color='green', alpha=0.1, label='Valence band')
    plt.axvspan(cbm, cbm + 0.5, color='red', alpha=0.1, label='Conduction band')

    # 图形设置
    plt.yticks(y_positions, labels, fontsize=8)
    plt.xlabel("Energy (eV)")
    plt.title(title)
    plt.xlim(vbm - 0.5, cbm + 0.5)
    plt.ylim(-1, len(fermi_dict))
    plt.legend(loc='upper right')
    plt.grid(axis='x', linestyle=':', alpha=0.5)
    plt.tight_layout()
    plt.show()


# 绘图
plot_fermi_levels(cu_fermi, "Fermi Levels for Cu Systems")
plot_fermi_levels(fe_fermi, "Fermi Levels for Fe Systems")
```
